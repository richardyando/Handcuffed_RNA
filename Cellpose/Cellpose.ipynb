# Cellpose
* conda create --name cellpose python=3.10
* conda activate cellpose
* install cellpose
    * gui
        * python -m pip install cellpose[gui]
    * terminal
        * python -m pip install cellpose
* install pytorch
    * conda install pytorch pytorch-cuda=11.8 -c pytorch -c nvidia

Run in the terminal.

python -m cellpose --verbose --dir confocal/ --pretrained_model cyto --chan 2 --use_gpu --do_3D --save_tif --no_npy --restore_type denoise_cyto3
